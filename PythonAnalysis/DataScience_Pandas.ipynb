# Pandas Indexing & Selection - loc & iloc

Label based selection

```python
subset = df.loc[35:50,["Category","Sales"]]
print(subset)
```

integer based selection

```python
subset =df.iloc[40:51,1:4]
print(subset)
```

boolena filter

```python
subset = df[df["Sales"]>1700]
print(subset)
```

assigmenet values or update through loc

```python
df.loc[df["Sales"]>1800 , "Region"]="Most Proitable sale"
print(df.loc[df["Sales"]>1800 , "Region"])
```
